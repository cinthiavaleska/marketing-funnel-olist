```python
import pandas as pd

df_leads = pd.read_csv("../data/raw/closed_deals.csv")
df_leads.head()
```

```python
# INVESTIGAR QUALIDADE DOS DADOS 
df_leads.info()

df_leads.isnull().sum()
```

```python
# VERIFICAR DUPLICATOS POR LINHAS

df_leads.duplicated().sum() 
```

```python
# TRATAR VALORES NULOS

# Colunas categóricas
colunas_categoricas = [
    'business_segment',
    'lead_type',
    'lead_behaviour_profile',
    'has_company',
    'has_gtin',
    'average_stock',
    'business_type'
]

for coluna in colunas_categoricas:
    df_leads[coluna] = df_leads[coluna].fillna('Não informado')

# Coluna numérica
mediana_catalogo = df_leads['declared_product_catalog_size'].median()

df_leads['declared_product_catalog_size'] = (
    df_leads['declared_product_catalog_size'].fillna(mediana_catalogo)
)

#VALIDAR NULOS TRATADOS

df_leads.isnull().sum()
```

```python
# CONVERTER WON_DATE PARA DATA

df_leads['won_date'] = pd.to_datetime(df_leads['won_date']).dt.date

df_leads.head(10)
```

```python
#EXPORTACAO DOS DADOS TRATADOS PARA A CAMADA SILVER


df_leads.to_csv(
    "../data/silver/closed_deals_clean.csv",
    index=False
)
```
