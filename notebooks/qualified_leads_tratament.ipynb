```python
import pandas as pd

df = pd.read_csv("../data/raw/qualified_leads.csv")
df.head()
```

```python
# INVESTIGAR QUALIDADE DOS DADOS 
df.info()

df.isnull().sum()
```

```python
# VERIFICAR DUPLICATOS POR LINHAS

df.duplicated().sum() 
```

```python
# MODIFICAR TYPE COLUNA DATE

df['first_contact_date'] = pd.to_datetime(df['first_contact_date'])

#VALIDAR CHANGE
df.info()
df.head()
```

```python
df['origin'].unique()

# TRATAR VALORES NULOS DA COLUNA ORIGIN

df['origin'] = df['origin'].fillna('unknown')

# VALIDAR

df['origin'].isnull().sum()
```

```python
#EXPORTACAO DOS DADOS TRATADOS PARA A CAMADA SILVER


df.to_csv(
    "../data/silver/qualified_leads_clean.csv",
    index=False
)
```
